```python
"""Explore Loss Functions

Structure:
    1. Imports, Variables, Functions
    2. Load Data
    3. Test Loss Functions
"""

# 1. Imports, Variables, Functions
# imports
import numpy as np
import pandas as pd, pickle, sys, os
import tempfile

# variables
file_do_sim =  "/aloy/home/ddalton/projects/BQ_diseases/outputs/do_similarity_results.pkl"
evaluated_dos = [
    "DOID:4",  # Cancer
    "DOID:162",  # Diabetes Mellitus
]

# functions



# 2. Load Data

# load Disease Ontology Lin Similarity amongst pairs!
results = pickle.load(open(file_do_sim, "rb"))
print(f"Loaded {len(results)} Disease Ontology pairs with similarity scores.")

df_do_ic_full = pd.DataFrame(
    results,
    columns=[
        "do1",
        "do2",
        "pair_sorted",
        "resnik",
        "lin",
        "jiang",
        "mica_node",
        "shortest_path_length"
    ]
)
print(f"DataFrame shape: {df_do_ic_full.shape}")
```

```python

# filter
df_do_ic = df_do_ic_full.query("do1 in @evaluated_dos and do2 in @evaluated_dos")
print(f"Filtered DataFrame shape: {df_do_ic.shape}")

# save filtered results in temporary csv file
# generate unique filename
# temp_dir = tempfile.gettempdir()
tempfile.tempdir = "/aloy/scratch/ddalton/tmp"
temp_file = tempfile.NamedTemporaryFile(suffix=".csv", delete=True) # if delete set to True it deletes the file immediately after closing

# save to csv
# Example: save DataFrame
df_do_ic.to_csv(temp_file.name, index=False)
    




# 3. Test Loss Functions




# at end of script - clean up temporary files
temp_file.close()
```
